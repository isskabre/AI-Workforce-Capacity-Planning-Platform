```python
# ============================================================================
# AI Workforce Capacity Planning Platform
# Implementation 29 — Production End-to-End Runtime Validation
#
# Cell 5: Enterprise Application Build Contract Validation
# ============================================================================

# ---------------------------------------------------------------------------
# Build enterprise application
# ---------------------------------------------------------------------------

application = application_factory.build()

assert application is not None, (
    "EnterpriseApplicationFactory.build() returned no application."
)


# ---------------------------------------------------------------------------
# Inspect actual application contract
# ---------------------------------------------------------------------------

application_type = type(application).__name__

public_contracts = tuple(
    sorted(
        name
        for name in dir(application)
        if not name.startswith("_")
    )
)

assert public_contracts, (
    "Built enterprise application exposes no public contracts."
)


# ---------------------------------------------------------------------------
# Inspect public contract values safely
# ---------------------------------------------------------------------------

contract_details = []

for contract in public_contracts:
    try:
        value = getattr(application, contract)
        value_type = type(value).__name__
        contract_details.append(
            (contract, value_type)
        )
    except Exception as exc:
        contract_details.append(
            (
                contract,
                f"<access failed: {type(exc).__name__}>",
            )
        )


# ---------------------------------------------------------------------------
# Validation result
# ---------------------------------------------------------------------------

print("Implementation 29 — Cell 5 PASSED")
print()

print("Enterprise application")
print(f"  ✓ Type      : {application_type}")
print(f"  ✓ Contracts : {len(public_contracts)}")
print()

for contract, value_type in contract_details:
    print(
        f"  ✓ {contract:<25} "
        f"[{value_type}]"
    )

print()
print("Enterprise application build contract is valid.")
```

```python
# ============================================================================
# AI Workforce Capacity Planning Platform
# Implementation 29 — Production End-to-End Runtime Validation
#
# Cell 6: Required Enterprise Service Composition Validation
# ============================================================================

# ---------------------------------------------------------------------------
# Resolve required enterprise service contracts
# ---------------------------------------------------------------------------

required_services = application_configuration.required_services

assert required_services, (
    "ApplicationConfiguration defines no required enterprise services."
)

assert len(required_services) == 6, (
    f"Expected 6 required services, found {len(required_services)}."
)


# ---------------------------------------------------------------------------
# Validate container registrations
# ---------------------------------------------------------------------------

registered_services = application.registered_services

assert registered_services, (
    "EnterpriseApplicationContainer contains no registered services."
)

missing_services = tuple(
    service_name
    for service_name in required_services
    if not application.contains(
        name=service_name,
    )
)

assert not missing_services, (
    "Required enterprise services are missing from the container: "
    f"{missing_services}"
)


# ---------------------------------------------------------------------------
# Resolve required enterprise services
# ---------------------------------------------------------------------------

resolved_services = application.resolve_required_services()

assert isinstance(resolved_services, dict), (
    "resolve_required_services() did not return the expected mapping."
)

assert set(resolved_services) == set(required_services), (
    "Resolved service names do not match the configured required services."
)


# ---------------------------------------------------------------------------
# Validate concrete service instances
# ---------------------------------------------------------------------------

for service_name in required_services:
    service = resolved_services.get(service_name)

    assert service is not None, (
        f"Required service '{service_name}' resolved to None."
    )


# ---------------------------------------------------------------------------
# Validate container state
# ---------------------------------------------------------------------------

assert application.service_count >= len(required_services), (
    "Container service count is smaller than the required service count."
)


# ---------------------------------------------------------------------------
# Validation result
# ---------------------------------------------------------------------------

print("Implementation 29 — Cell 6 PASSED")
print()

print("Enterprise service composition")
print(f"  ✓ Required services   : {len(required_services)}")
print(f"  ✓ Registered services : {len(registered_services)}")
print(f"  ✓ Container count     : {application.service_count}")
print()

for service_name in required_services:
    service = resolved_services[service_name]

    print(
        f"  ✓ {service_name:<25} "
        f"[{type(service).__name__}]"
    )

print()
print("All required enterprise services are registered and resolvable.")
```

```python
# ============================================================================
# AI Workforce Capacity Planning Platform
# Implementation 29 — Production End-to-End Runtime Validation
#
# Cell 7: Enterprise Runner Lifecycle Validation
# ============================================================================

from src.runner import (
    EnterprisePlatformRunner,
    RunnerConfiguration,
    RunnerStatus,
)


# ---------------------------------------------------------------------------
# Create isolated runner instance
# ---------------------------------------------------------------------------

lifecycle_configuration = RunnerConfiguration(
    auto_startup=False,
    auto_shutdown=False,
)

lifecycle_runner = EnterprisePlatformRunner(
    configuration=lifecycle_configuration,
)

runner_service = lifecycle_runner.service


# ---------------------------------------------------------------------------
# Validate initial state
# ---------------------------------------------------------------------------

assert runner_service.status is RunnerStatus.CREATED, (
    "Runner service did not initialize in CREATED status."
)

assert runner_service.is_created, (
    "Runner service did not report is_created=True."
)

assert not runner_service.is_running, (
    "Runner service incorrectly reports RUNNING before startup."
)

assert not runner_service.is_stopped, (
    "Runner service incorrectly reports STOPPED before startup."
)


# ---------------------------------------------------------------------------
# Execute startup
# ---------------------------------------------------------------------------

startup_result = runner_service.start()

assert startup_result.succeeded, (
    "Enterprise runner startup did not succeed."
)

assert startup_result.descriptor.status is RunnerStatus.RUNNING, (
    "Startup result did not return RUNNING status."
)

assert runner_service.status is RunnerStatus.RUNNING, (
    "Runner service did not transition to RUNNING."
)

assert runner_service.is_running, (
    "Runner service did not report is_running=True after startup."
)


# ---------------------------------------------------------------------------
# Execute graceful shutdown
# ---------------------------------------------------------------------------

shutdown_result = runner_service.stop()

assert shutdown_result.succeeded, (
    "Enterprise runner shutdown did not succeed."
)

assert shutdown_result.descriptor.status is RunnerStatus.STOPPED, (
    "Shutdown result did not return STOPPED status."
)

assert runner_service.status is RunnerStatus.STOPPED, (
    "Runner service did not transition to STOPPED."
)

assert runner_service.is_stopped, (
    "Runner service did not report is_stopped=True after shutdown."
)


# ---------------------------------------------------------------------------
# Validation result
# ---------------------------------------------------------------------------

print("Implementation 29 — Cell 7 PASSED")
print()

print("Enterprise runner lifecycle")
print("  ✓ Initial status : CREATED")
print("  ✓ Startup        : RUNNING")
print("  ✓ Shutdown       : STOPPED")
print()

print(f"  ✓ Startup message  : {startup_result.message}")
print(f"  ✓ Shutdown message : {shutdown_result.message}")
print(f"  ✓ Exit code        : {shutdown_result.exit_code}")

print()
print("Enterprise runner lifecycle is valid.")
```

```python
# ============================================================
# AI Workforce Capacity Planning Platform
# Implementation 30.2 — Enterprise Decision Workspace Enhancement
# Part 2 — Manager Action View Validation
# ============================================================

from pathlib import Path
import ast

APP_PATH = Path(
    "/Workspace/Users/isskabre@gmail.com/"
    "AI_Workforce_Capacity_Planning_Platform_Git/"
    "deployment/databricks_app/app.py"
)

source = APP_PATH.read_text(
    encoding="utf-8"
)

# ------------------------------------------------------------
# 1. Python syntax validation
# ------------------------------------------------------------

compile(
    source,
    str(APP_PATH),
    "exec",
)

# ------------------------------------------------------------
# 2. Implementation identity preserved
# ------------------------------------------------------------

assert (
    "Implementation 30.2 — "
    "Enterprise Decision Workspace Enhancement"
    in source
)

assert '"implementation": "30.2"' in source


# ------------------------------------------------------------
# 3. Manager Action View exists
# ------------------------------------------------------------

required_manager_view_tokens = (
    "render_manager_action_view",
    'st.subheader("Manager Action View")',
    'label="Capacity Gap"',
    'label="Recommended Action"',
    'label="Decision Priority"',
    'label="Recommended Associates"',
    'label="Overtime Hours"',
    'label="Optimization Status"',
    "**Decision Rationale**",
)

for token in required_manager_view_tokens:
    assert token in source, (
        f"Missing Part 2 manager-view contract: {token}"
    )


# ------------------------------------------------------------
# 4. Manager view uses canonical API payload fields
# ------------------------------------------------------------

required_api_fields = (
    "associate_gap",
    "staffing_recommendation",
    "optimization_priority",
    "optimization_status",
    "recommended_associates",
    "overtime_hours",
    "recommendation_rationale",
    "rationale",
    "decision_rationale",
)

for field_name in required_api_fields:
    assert field_name in source, (
        f"Missing canonical API field: {field_name}"
    )


# ------------------------------------------------------------
# 5. No UI-side decision thresholds introduced
# ------------------------------------------------------------

tree = ast.parse(
    source,
    filename=str(APP_PATH),
)

manager_function = None

for node in tree.body:
    if (
        isinstance(node, ast.FunctionDef)
        and node.name == "render_manager_action_view"
    ):
        manager_function = node
        break

assert manager_function is not None

# The manager view should present API outputs, not calculate decisions.
for node in ast.walk(manager_function):

    if isinstance(node, ast.If):
        condition_text = ast.unparse(
            node.test
        )

        assert "associate_gap" not in condition_text
        assert "recommended_associates" not in condition_text
        assert "overtime_hours" not in condition_text
        assert "optimization_priority" not in condition_text
        assert "staffing_recommendation" not in condition_text


# ------------------------------------------------------------
# 6. No direct domain-layer imports introduced
# ------------------------------------------------------------

forbidden_direct_imports = (
    "src.planning",
    "src.staffing",
    "src.overtime",
    "src.optimization",
    "src.workforce",
    "src.orchestration",
)

imported_modules = []

for node in ast.walk(tree):

    if isinstance(node, ast.Import):
        imported_modules.extend(
            alias.name
            for alias in node.names
        )

    elif isinstance(node, ast.ImportFrom):
        if node.module:
            imported_modules.append(
                node.module
            )

for module_name in imported_modules:
    assert not any(
        module_name == forbidden
        or module_name.startswith(
            f"{forbidden}."
        )
        for forbidden in forbidden_direct_imports
    ), (
        "Manager Action View must not bypass "
        f"the Enterprise API boundary: {module_name}"
    )


# ------------------------------------------------------------
# 7. Enterprise API decision boundary preserved
# ------------------------------------------------------------

assert "ENDPOINT_DECISION" in source
assert "OPERATION_CREATE_DECISION" in source
assert "decision_response = call_api(" in source
assert "path=ENDPOINT_DECISION" in source
assert "method=HTTP_METHOD_POST" in source
assert "operation=OPERATION_CREATE_DECISION" in source


# ------------------------------------------------------------
# 8. Scenario presets preserved
# ------------------------------------------------------------

for scenario_name in (
    "Balanced Capacity",
    "Capacity Surplus",
    "Moderate Shortage",
    "Severe / Persistent Shortage",
):
    assert scenario_name in source

assert '"available_associates": 59' in source
assert '"available_associates": 60' in source
assert '"available_associates": 50' in source


# ------------------------------------------------------------
# 9. Existing result presentation preserved
# ------------------------------------------------------------

assert 'label="Workflow Status"' in source
assert 'label="Forecast Confidence"' in source
assert 'label="Expected Order Lines"' in source
assert 'label="Available Associates"' in source
assert 'label="Required Associates"' in source
assert 'label="Associate Gap"' in source
assert 'label="Staffing Recommendation"' in source
assert 'label="Optimization Priority"' in source


# ------------------------------------------------------------
# 10. Unsupported confidence concepts remain removed
# ------------------------------------------------------------

assert (
    'label="Recommendation Confidence"'
    not in source
)

assert (
    'label="Decision Confidence"'
    not in source
)

assert '"recommendation_confidence"' not in source


# ------------------------------------------------------------
# Validation summary
# ------------------------------------------------------------

print(
    "Implementation 30.2 — Part 2 Validation PASSED"
)

print()

print("✓ app.py compiled successfully")
print("✓ Implementation 30.2 identity preserved")
print("✓ Manager Action View contracts confirmed")
print("✓ Canonical Enterprise API fields are used")
print("✓ No UI-side workforce decision thresholds detected")
print("✓ No direct domain-layer imports introduced")
print("✓ Enterprise API decision boundary preserved")
print("✓ Scenario presets preserved")
print("✓ Existing result presentation preserved")
print("✓ Unsupported confidence concepts remain removed")

print()

print(
    "Implementation 30.2 Part 2 is ready "
    "for live Databricks App validation."
)
```
