# Implementation 02 — Public Dataset Evaluation and Selection

## Business Objective

Select a public dataset that can support the development of an AI Workforce
Capacity Planning prototype for a distribution center.

The selected dataset must be suitable for:

- Daily workload forecasting
- Traditional machine learning model comparison
- LSTM experimentation
- Capacity planning simulation
- Overtime recommendation
- Stakeholder demonstration

## Core Decision

The project will forecast future operational workload rather than directly
predict overtime.

Forecasted workload will later be converted into:

1. Required labor hours
2. Available labor capacity
3. Capacity gap
4. Recommended overtime

## Dataset Evaluation Criteria

Each candidate dataset will be evaluated using the following criteria:

| Criterion | Weight |
|---|---:|
| Distribution-center similarity | 25% |
| Daily time-series availability | 20% |
| Historical depth | 15% |
| Workload target quality | 15% |
| Feature richness | 10% |
| Traditional ML suitability | 5% |
| LSTM suitability | 5% |
| Documentation and licensing | 5% |

The final dataset score will be calculated as a weighted score out of 10.

## Candidate 1 — SMART Supply Chain

### Dataset purpose

The SMART Supply Chain dataset contains transactional supply-chain information
related to orders, customers, products, shipping, delivery, and distribution.

### Evaluation question

Does this dataset contain a sufficiently long and continuous daily time series
that can be used to forecast distribution-center workload?

### Initial findings

| Attribute | Finding |
|---|---|
| Domain | Supply chain and logistics |
| Business process | Orders, shipping, delivery, products, and customers |
| Operational similarity | High |
| Primary data grain | Transaction or order-line level |
| Potential workload measure | Daily orders, daily order lines, or daily shipped units |
| Forecasting suitability | To be verified |
| Main risk | Transactional data may not provide enough continuous daily history |

```python
from pyspark.sql.types import (
    StringType,
    StructField,
    StructType,
)

smart_facts_schema = StructType(
    [
        StructField("attribute", StringType(), False),
        StructField("finding", StringType(), False),
        StructField("status", StringType(), False),
    ]
)

smart_supply_chain_facts = [
    (
        "Domain",
        "Supply chain and logistics",
        "Confirmed",
    ),
    (
        "Business processes",
        "Orders, products, customers, shipping, delivery, and distribution",
        "Confirmed",
    ),
    (
        "Primary grain",
        "Transactional order or order-line records",
        "Confirmed",
    ),
    (
        "Potential workload target",
        "Daily order count, order-line count, or shipped units",
        "Candidate",
    ),
    (
        "Daily continuity",
        "Must be verified from the source data",
        "Pending",
    ),
    (
        "Historical depth",
        "Must be verified from the source data",
        "Pending",
    ),
    (
        "LSTM suitability",
        "Depends on the number of continuous daily observations",
        "Pending",
    ),
]

smart_supply_chain_facts_df = spark.createDataFrame(
    smart_supply_chain_facts,
    schema=smart_facts_schema,
)

display(smart_supply_chain_facts_df)
```
